```python
menu = {
    'apple' : 1,
    'granola bar' : 2,
    'pretzees' : 2,
    'juice' : 3,
}

order = ['granola bar', 'apple', 'pizza']
```

```python
total = 0
discount = 0

for item in order:
    if item in menu:
        total = total + menu[item]
    else:
        print('Item not found in the menu: ', item)
```

```python
if total > 5:
    discount = 1
    total = total - discount
else:
    print('Not apply discount')

print('Order item: ', order)
print('Discount: ', discount)
print('Order total: ', total)
```

```python
def caculate_total(items, menu, discount=0):
    total = 0
    for item in items:
        if item in menu:
            total += menu[item]
        else:
            print('Item not in menu: ', item)
    print('Total: ', total)
    return {'order total' : total,
            'num items' : len(items)
            }
```

```python
total_price = caculate_total(['apple', 'juice', 'granola bar'], menu)
```
