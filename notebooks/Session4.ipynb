```python
import numpy as np
import pandas as pd
```

```python
scores = np.array([85,90, 78, 92, 88])

print(scores.size)
```

```python
print(scores[-2:])
```

```python
scores[0] = 75
print(scores)
```

```python
scores = scores + 5
print(scores)
```

```python
scores = scores / 100
print(scores)

scores = scores ** 2
print(scores)
```

```python
print('Sum of scores:', np.sum(scores))
print('Average of scores:', scores.mean())
print('Minimum score:', scores.min())
print('Maximum score:', scores.max())
```
