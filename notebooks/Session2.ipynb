## Conditional Logic

```python
money = 3
snack_price = 3

if money >= snack_price:
    print("You can buy "
        "the snack")
    print("Horray!")
elif money > 0:
    print("You don't have enough money to buy the snack.")
elif money > 2:
    print("You're getting close!")
else:
    print("You cannot buy the snack.")
```
