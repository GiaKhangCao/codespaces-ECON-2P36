```python
import pandas as pd
```

```python
fred_endpoint = "https://fred.stlouisfed.org/graph/fredgraph.csv?id="
cpi_meonic = "CPIAUCSL"
un_menomic = "UNRATE"
```

```python
fred_endpoint + cpi_meonic
```

```python
un_raw = pd.read_csv(fred_endpoint + un_menomic)
cpi_raw = pd.read_csv(fred_endpoint + cpi_meonic)
```

```python
un_raw.head()
```

```python
cpi_raw.head()
```

```python
un_raw.dtypes
```

```python
print('Missing values in each raw file:')
print('UNRATE (unemployment rate):', un_raw.isna().sum().to_dict())
print('CPIAUCSL (CPI level):', cpi_raw.isna().sum().to_dict())
```

```python
unemployment = un_raw.rename(
    columns={
        'observation_date' : 'date',
        'UNRATE' : 'unemployment_rate',
    }
).copy() #Create deep copyy - ensures unemployemnt is SPERATE from un_raw

cpi = cpi_raw.rename(
    columns={
        'observation_date' : 'date',
        'CPIAUCSL' : 'cpi',
    }
).copy()
```

```python
unemployment.head()
```

```python
unemployment['date'] = pd.to_datetime(unemployment['date'])
cpi['date'] = pd.to_datetime(cpi['date'])

unemployment = unemployment.sort_values('date')
cpi = cpi.sort_values('date')
```

```python
unemployment.tail()
```

```python
print('Unemployment dates - all unique:', unemployment['date'].is_unique)
print('CPI dates - all unique:', cpi['date'].is_unique)
print('Unemployment dates - sample size:', len(unemployment))
print('CPI- sample size:', len(cpi))
```

```python
print('Unemployment start is', unemployment['date'].min(),
      ' and end is ', unemployment['date'].max())
print('CPI start is ', cpi['date'].min(),
      ' and end is ', cpi['date'].max())
```
