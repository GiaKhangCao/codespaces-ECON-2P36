# 2. Variables: giving values their names

Variable is a name that refers to a value. To assign a value to a variable , we use the = sign

```python
student_name = "Alice"
age = 20
has_passed = False

print(student_name)
print(age)
print(has_passed)
```

```python
city = "Ottawa"
tempeature = 999999999
Sunny = True

print(city)
print(tempeature)
print(Sunny)
```

```python
print(type(age))
print(type(has_passed))
print(type(student_name))
```

```python
a = 17
b = 5
print(a + b) # addition
print(a - b) # subtraction
print(a * b) # multiplication
print(a / b) # division
print(a % b) # modulus
print(a ** b) # exponentiation
print(a ** (1/2)) # square root
print(a // b) # floor division
```

```python
print((a//b) + (a%b)) # floor division + modulus
```

# Updating Variables

```python
score = 10
print(score)
score = score + 5
print(score)

score += 5
score *= 2
print(score)
```

```python
ticket_price = 12
number_of_tickets = 3
snack_fee = 5
total = (ticket_price + snack_fee) * number_of_tickets
print(total)
```
